```python
import pandas as pd

# CSV file load
df = pd.read_csv("netflix_titles.csv")

# First 5 rows
print(df.head())

# Dataset info
print(df.info())
```

# **Check Missing Values**

```python
# Missing value check
print(df.isnull().sum())

# Example: Missing values fill
df['director'] = df['director'].fillna("Unknown")
df['cast'] = df['cast'].fillna("Not Available")

df = df.dropna(subset=['date_added'])
```

```python
# Duplicate check
print(df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()
```

```python
# lowercase
df['country'] = df['country'].str.strip().str.title()

# Rating column check
print(df['rating'].unique())
```

```python
# Convert date_added column to datetime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
# clean column namee
df.columns = df.columns.str.lower().str.replace(" ", "_")
```

```python
# Save clean Dataset
df.to_csv("netflix_cleaned.csv", index=False)
```

# **3.** Difference between dropna() and fillna() in Pandas?
dropna() → removes rows/columns with missing values.

fillna() → replaces missing values with a specific value (mean, median, mode, or custom).

# **4**. What is outlier treatment and why is it important?
Outliers are extreme values that deviate from the rest of the data.

Treatment: detect (IQR, Z-score), then remove or cap them.

Importance: prevents skewed results and improves model accuracy.

# **5.** Explain the process of standardizing data.
Make data consistent across the dataset.

Examples:

Text → same format (“M”, “Male” → “Male”)

Dates → same format (YYYY-MM-DD)

Numeric scaling → standardize to a common range or distribution.
