```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Titanic dataset load (from seaborn)
df = sns.load_dataset("titanic")
```

Data Overview (.describe(), .info(), .value_counts())

```python


# Summary statistics
print(df.describe(include="all"))

# Info about dataset
print(df.info())

# Value counts example (Survived column)
print(df['survived'].value_counts())
```

Pairplot & Heatmap (Relationships)

```python

# Pairplot for selected numerical columns
sns.pairplot(df[['age', 'fare', 'survived']], hue="survived")
plt.show()

# Correlation heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

Histograms, Boxplots, Scatterplots

```python

# Histogram of Age
df['age'].hist(bins=30)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")
plt.show()

# Boxplot of Fare by Class
sns.boxplot(x='class', y='fare', data=df)
plt.title("Fare distribution by Class")
plt.show()

# Scatterplot: Age vs Fare
sns.scatterplot(x='age', y='fare', hue='survived', data=df)
plt.title("Age vs Fare (colored by Survival)")
plt.show()
```

# ***Write observations for each visual***

# 1. describe(include="all") → Summary Statistics

Observation:

Average passenger age ≈ 29 years.

Most fares are cheap, but some are very high (outliers).

Survival rate is about 38% (mean of survived ≈ 0.38).

Most passengers are in 3rd class, fewer in 1st class.

# 2. info() → Dataset Info

Observation:

Dataset has both numeric (age, fare, pclass) and categorical (sex, embarked, class) columns.

Some columns have missing values (e.g. age, cabin).

Total rows ≈ 891.

# 3. value_counts() → Example: survived

Observation:

More people died (≈ 62%) than survived (≈ 38%).

Male passengers are more in number than females.

3rd class has the most passengers.

# 4. Pairplot (sns.pairplot)

Observation:

Survivors are often passengers with higher fare (upper side).

1st class (high fare) passengers had better survival chances.

Younger people, especially children, had more survival compared to elderly.

# 5. Heatmap (sns.heatmap)

Observation:

Strong negative relation: Pclass vs Fare (higher class → higher fare).

Survival is slightly positively related with fare (rich passengers survived more).

Age has very weak relation with survival.

# 6. Histogram (df['age'].hist())

Observation:

Most passengers are between 20–40 years old.

Few children under 10 and fewer elderly (60+).

Distribution is slightly right-skewed (more younger people).

# 7. Boxplot (sns.boxplot(x='class', y='fare'))

Observation:

Median fare is lowest in 3rd class, highest in 1st class.

1st class has many outliers with extremely high fares.

Fare clearly separates the social/economic groups.

# 8. Scatterplot (sns.scatterplot(x='age', y='fare', hue='survived'))

Observation:

Survivors are more in high-fare region.

Passengers with very low fares mostly did not survive.

Age does not show a clear pattern, but some children survived more often.

# ***Summary***

More people died (about 62%) than survived (about 38%).

3rd class passengers were the majority, but they had the lowest survival rate.

1st class passengers paid higher fares and had a better chance of survival.

Women and children had higher survival compared to men.

Most passengers were 20–40 years old.

Age alone did not decide survival, but young children survived more often.

Higher fare passengers survived more (likely because they were in better classes).
